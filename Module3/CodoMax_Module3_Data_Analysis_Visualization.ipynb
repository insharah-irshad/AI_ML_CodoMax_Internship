```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
data = {
    "Date": [
        "2026-01-05",
        "2026-01-12",
        "2026-01-20",
        "2026-02-03",
        "2026-02-15",
        "2026-02-25",
        "2026-03-05",
        "2026-03-18",
        "2026-03-27",
        "2026-04-10"
    ],
    "Product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Laptop",
        "Headphones",
        "Mouse",
        "Keyboard",
        "Laptop",
        "Headphones",
        "Mouse"
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Accessories",
        "Electronics",
        "Accessories",
        "Accessories",
        "Accessories",
        "Electronics",
        "Accessories",
        "Accessories"
    ],
    "Quantity": [
        2, 10, 5, 3, 8, 15, 7, 2, 6, 12
    ],
    "Price": [
        80000, 1500, 3000, 80000, 5000,
        1500, 3000, 80000, 5000, 1500
    ]
}

df = pd.DataFrame(data)

df["Sales"] = df["Quantity"] * df["Price"]

df.head()
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Statistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Check for duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows if any
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
```

```python
display(df)
```

```python
# Basic sales analysis

total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
highest_sale = df["Sales"].max()
lowest_sale = df["Sales"].min()

print("Total Sales:", total_sales)
print("Average Sale:", round(average_sales, 2))
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
# Sales by category

category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
# Sales by product

product_sales = df.groupby("Product")["Sales"].sum()

print(product_sales)
```

```python
best_product = product_sales.idxmax()
best_product_sales = product_sales.max()

print("Best-Selling Product:", best_product)
print("Sales:", best_product_sales)
```

```python
# Sort data by date
df = df.sort_values("Date")

plt.figure(figsize=(10, 5))

plt.plot(df["Date"], df["Sales"], marker="o")

plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

The line chart shows how sales changed over time. It helps identify increases and decreases in sales across different dates.

```python
# Sales distribution by category

category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Category")

plt.show()
```

The pie chart shows the percentage distribution of total sales across different product categories. It provides a clear view of each category's contribution to overall sales.
